```python
cod=str(input("introduceti codul produsului"))
cifre=0
caract=0
P=0
if cod[0]=='P':
    P=1
for i in range(len(cod)):
    if cod[i] in "123456789":
        cifre +=1
    elif cod[i] in "!@#$%^&*()_+":
        caract +=1
        i=len(cod)
litere=len(cod)-cifre
lungime=len(cod)
ok=0
if lungime>=6 and (litere>=2 and cifre>=2) and caract==0:
    ok=1
if P==1:
    cod='A'+cod[1:]
if ok==1:
    print("Codul este valid")
else:
    print("Codul nu este valid")
print(cod)
```
